## Exercise 1: `generate_draft` Function

**Objective**:
Write a function called `generate_draft` that takes in a string topic and uses a language model to generate a complete draft essay.

**Inputs**:

* `topic` (str): The essay topic.
* `model` (str, optional): The model identifier to use. Defaults to `"openai:gpt-4o"`.

**Output**:

* A string representing the full draft of the essay.

The setup for calling the LLM using the aisuite library is already provided. Focus on crafting the prompt content. You can reference this setup in later exercises to understand how to interact with the library effectively.


response = {
    "choices": [

            "message": {

            }
        }
    ],
    "model": "gpt-4o",

        "prompt_tokens": 100,
        "completion_tokens": 500

}

```python
# GRADED FUNCTION: generate_draft
# 定义一个函数 generate_draft，用于根据给定的作文主题生成一篇完整草稿

def generate_draft(
    topic: str,                         # 输入参数：作文主题，类型为字符串
    model: str = "openai:gpt-4o"        # 输入参数：指定使用的模型，默认使用 gpt-4o
) -> str:                               # 函数最终返回一个字符串，也就是生成的作文内容

    ### START CODE HERE ###

    # 构造发送给大语言模型的 Prompt
    # 使用 f-string，可以把用户输入的 topic 动态插入到提示词中
    prompt = f"""
    Write a complete, well-structured essay on the following topic:

    Topic: {topic}

    The essay should include:
    - an introduction
    - several body paragraphs
    - a conclusion

    Make the essay clear, coherent, and detailed.
    """

    ### END CODE HERE ###

    # 调用大语言模型 API，创建一次聊天补全请求
    response = CLIENT.chat.completions.create(

        # 指定要使用的模型
        # 这里的 model 来自函数参数
        model=model,

        # messages 是发送给聊天模型的消息列表
        # role="user" 表示这条消息是用户发送的
        # content=prompt 表示真正发送给模型的内容是上面构造好的 prompt
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],

        # temperature 控制模型生成结果的随机性
        # 数值越高，输出通常越多样、越有创造性
        # 数值越低，输出通常越稳定、确定
        temperature=1.0,
    )

    # response 是模型返回的完整响应对象
    # choices[0]：取第一个生成结果
    # message：取得模型生成的消息
    # content：取得消息中的真正文本内容
    #
    # 最终返回生成的 essay draft
    return response.choices[0].message.content
```

## Exercise 2: `reflect_on_draft` Function

**Objective**:
Write a function called `reflect_on_draft` that takes a previously generated essay draft and uses a language model to provide constructive feedback.

**Inputs**:

* `draft` (str): The essay text to reflect on.
* `model` (str, optional): The model identifier to use. Defaults to `"openai:o4-mini"`.

**Output**:

* A string with feedback in paragraph form.

**Requirements**:

* The feedback should be critical but constructive.
* It should address issues such as structure, clarity, strength of argument, and writing style.
* The function should send the draft to the model and return its response.

You do **not** need to rewrite the essay at this step—just analyze and reflect on it.


f""" ... {draft} ... """

```python
# GRADED FUNCTION: reflect_on_draft

def reflect_on_draft(
    draft: str,                         # 输入：之前生成的作文草稿
    model: str = "openai:o4-mini"       # 输入：指定使用的模型，默认使用 o4-mini
) -> str:                               # 输出：返回字符串，也就是模型生成的反馈内容

    ### START CODE HERE ###

    # 构造 reflection prompt
    # f"""...""" 是多行 f-string，
    # 这样可以把变量 draft 的真实内容插入到 prompt 中
    prompt = f"""
    Please critically but constructively reflect on the following essay draft.

    Focus on the following aspects:
    - structure
    - clarity
    - strength of argument
    - writing style

    Provide concise feedback in paragraph form.
    Do not rewrite the essay.

    Draft:
    {draft}
    """

    ### END CODE HERE ###

    # 调用大语言模型
    # 把上面构造好的 prompt 作为 user message 发送给模型
    response = CLIENT.chat.completions.create(

        # 使用函数参数中指定的模型
        model=model,

        # messages 是发送给聊天模型的消息
        # role="user" 表示这条消息来自用户
        # content=prompt 表示真正发送的是上面的 reflection prompt
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],

        # 控制模型输出的随机性
        # 值越高，输出通常越多样
        temperature=1.0,
    )

    # 从完整的 response 中取出模型真正生成的反馈文本
    # choices[0]：第一个生成结果
    # message：模型生成的消息
    # content：消息中的文本内容
    return response.choices[0].message.content
```

## Exercise 3: `revise_draft` Function

**Objective**:
Implement a function called `revise_draft` that improves a given essay draft based on feedback from a reflection step.

**Inputs**:

* `original_draft` (str): The initial version of the essay.
* `reflection` (str): Constructive feedback or critique on the draft.
* `model` (str, optional): The model identifier to use. Defaults to `"openai:gpt-4o"`.

**Output**:

* A string containing the revised and improved essay.

**Requirements**:

* The revised draft should address the issues mentioned in the feedback.
* It should improve clarity, coherence, argument strength, and overall flow.
* The function should use the feedback to guide the revision, and return only the final revised essay.

In this final exercise, you'll also need to manage the call to the LLM using the CLIENT, as you've practiced in previous exercises.

```python
# GRADED FUNCTION: revise_draft

def revise_draft(
    original_draft: str,
    reflection: str,
    model: str = "openai:gpt-4o"
) -> str:

    ### START CODE HERE ###

    # 构造 prompt：
    # 把原始 draft 和 reflection 都提供给模型，
    # 让模型根据 feedback 生成修改后的版本
    prompt = f"""
    Here is the original draft:

    {original_draft}

    Here is the reflection / feedback:

    {reflection}

    Please revise the original draft according to the feedback above.
    Generate a complete revised version of the essay.
    """

    # 调用 LLM
    response = CLIENT.chat.completions.create(
        model=model,

        # messages 必须是一个 list
        # list 里面的每一个元素都是一个 message 字典
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    ### END CODE HERE ###

    # 返回修改后的作文文本
    return response.choices[0].message.content
```

```python
# Test your code!
unittests.test_revise_draft(revise_draft)
```
